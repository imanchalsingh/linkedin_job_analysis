```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
jobs = pd.read_csv('cleaned_jobs.csv', sep=',', na_values=[''], quotechar='"')
```

```python
jobs.head()
```

```python
jobs.shape
```

```python
plt.scatter(jobs['application_type'], jobs['views'])
plt.xlabel('application_type')
plt.ylabel('views')
plt.show()
```

```python
work_type_counts = jobs['work_type'].value_counts()

plt.bar(work_type_counts.index, work_type_counts.values)
plt.xlabel('work_type')
plt.ylabel('Count')
plt.show()
```
