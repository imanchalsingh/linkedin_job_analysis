```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
jobs = pd.read_csv('jobs.csv', sep=',', na_values=[''], quotechar='"')
```

```python
jobs.columns
```

```python
jobs['remote_allowed'] = jobs['remote_allowed'].fillna("No",axis=0)
jobs['remote_allowed'] = jobs['remote_allowed'].replace(1,"Yes")
```

```python
jobs['applies'] = jobs['applies'].fillna("0")
jobs['applies'] = jobs['applies'].astype(int)
```

```python
jobs['posting_date'] = pd.to_datetime(jobs['original_listed_time'], unit='ms').dt.date
jobs['posting_time'] = pd.to_datetime(jobs['original_listed_time'], unit='ms').dt.time
```

```python
jobs['expiry_date'] = pd.to_datetime(jobs['expiry'], unit='ms').dt.date
jobs['expiry_time'] = pd.to_datetime(jobs['expiry'], unit='ms').dt.time
```

```python
jobs.drop('expiry',axis=1, inplace=True)
```

```python
jobs.drop('original_listed_time',axis=1, inplace=True)
```

```python
jobs['experience_level'] = jobs['formatted_experience_level'].fillna("No-Experience")
```

```python
jobs.drop('formatted_experience_level',axis=1, inplace=True)
```

```python
counts = jobs['compensation_type'].value_counts()
```

```python
counts
```

```python
jobs.shape
```

```python
jobs['application_type'] = jobs['application_type'].replace('ComplexOnsiteApply',"Onsite")
jobs['application_type'] = jobs['application_type'].replace('OffsiteApply',"Offsite")
jobs['application_type'] = jobs['application_type'].replace('SimpleOnsiteApply',"Onsite")
jobs['application_type'] = jobs['application_type'].replace('UnknownApply',"Offsite")
```

```python
count = jobs['application_type'].value_counts()
count
```

```python
com_count = jobs['compensation_type'].value_counts()
com_count
```

```python
curr_count = jobs['currency'].value_counts()
curr_count
```

```python
jobs['currency'] = jobs['currency'].fillna("NAN")
```

```python
jobs['max_salary'] = pd.to_numeric(jobs['max_salary'], errors='coerce').fillna(0).astype(float).astype(int)

jobs['min_salary'] = pd.to_numeric(jobs['min_salary'], errors='coerce').fillna(0).astype(float).astype(int)
```

```python
jobs['normalized_salary'] = pd.to_numeric(jobs['normalized_salary'], errors='coerce').fillna(0).astype(float).astype(int)
```

```python
jobs.drop('listed_time',axis=1,inplace=True)
```

```python
jobs['views'] = pd.to_numeric(jobs['views'], errors='coerce').fillna(0).astype(float).astype(int)
```

```python
jobs.dropna(inplace=True)
```

```python
jobs = jobs.to_csv('cleaned_jobs.csv', sep=',', index=False)
```
